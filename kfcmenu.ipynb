### 1. Load the Dataset and Perform Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV data
file_path = '/content/nutrition-booklet.csv'
df = pd.read_csv(file_path)

# Display basic information
print("--- Dataset Info ---")
display(df.info())

print("\n--- First 5 Rows ---")
display(df.head())

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- Summary Statistics ---")
display(df.describe(include='all'))
```

### 2. Preprocess Data, Build, and Train Decision Tree Classifier

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder

# --- 1. Clean the Data ---
# Drop completely empty columns
df_clean = df.dropna(how='all', axis=1)

# Drop rows 0 and 1 which contain metadata/units ('g', 'Kcal', etc.)
df_clean = df_clean.iloc[2:].reset_index(drop=True)

# Drop rows where 'Products' is missing, as it is essential
df_clean = df_clean.dropna(subset=['Products'])

# For demonstration, let's create a classification target.
# Let's predict 'Allergens' or create a binned class of 'Total Energy'.
# Clean 'Total Energy' column and convert to numeric
df_clean['Total Energy'] = pd.to_numeric(df_clean['Total Energy'].astype(str).str.replace(r'[^\d.]', '', regex=True), errors='coerce')
# Fill NaN values in energy with the median
df_clean['Total Energy'] = df_clean['Total Energy'].fillna(df_clean['Total Energy'].median())

# Create a target: High Energy vs Low Energy (split by median)
median_energy = df_clean['Total Energy'].median()
df_clean['Energy_Class'] = np.where(df_clean['Total Energy'] >= median_energy, 'High', 'Low')

# Target is 'Energy_Class', we drop 'Total Energy' to avoid data leakage
target_column = 'Energy_Class'
X = df_clean.drop(columns=['Energy_Class', 'Total Energy'])
y = df_clean[target_column]

# Fill remaining missing values with 'Unknown' for categorical features
X = X.fillna('Unknown')

# --- 2. Encode Categorical Features ---
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

# Encode target
le_target = LabelEncoder()
y_encoded = le_target.fit_transform(y)

# --- 3. Split data and train Decision Tree ---
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

clf = DecisionTreeClassifier(max_depth=5, random_state=42)
clf.fit(X_train, y_train)

# --- 4. Evaluate and Report ---
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}\n")

print("--- Classification Report ---")
print(classification_report(y_test, y_pred, target_names=le_target.classes_))
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
import re

# --- 1. Clean the Data ---
# Drop completely empty columns
df_advanced = df.dropna(how='all', axis=1).iloc[2:].reset_index(drop=True)
df_advanced = df_advanced.dropna(subset=['Products'])

# Define the lists of columns based on their roles
categorical_cols = ['Products', 'Allergens', 'MSG Info', 'Caffeine and Artificial Sweetene\nr Info']
# Any column not explicitly categorical and not target-related is a candidate for numerical parsing
numeric_candidates = [col for col in df_advanced.columns if col not in categorical_cols + ['Total Energy', 'Energy_Class']]

# Clean and convert the numerical candidates
for col in numeric_candidates:
    # Clean strings (remove spaces, units, non-numeric chars) and convert to float
    df_advanced[col] = pd.to_numeric(df_advanced[col].astype(str).str.replace(r'[^\d.]', '', regex=True), errors='coerce')
    # Impute missing numerical values with column median
    df_advanced[col] = df_advanced[col].fillna(df_advanced[col].median())

# Re-create Target based on 'Total Energy'
df_advanced['Total Energy'] = pd.to_numeric(df_advanced['Total Energy'].astype(str).str.replace(r'[^\d.]', '', regex=True), errors='coerce')
median_energy = df_advanced['Total Energy'].median()
df_advanced['Energy_Class'] = np.where(df_advanced['Total Energy'] >= median_energy, 'High', 'Low')

# Define X and y
X_adv = df_advanced.drop(columns=['Energy_Class', 'Total Energy'])
y_adv = df_advanced['Energy_Class']

# Handle missing values for categorical features before one-hot encoding
for col in categorical_cols:
    if col in X_adv.columns:
        X_adv[col] = X_adv[col].fillna('Unknown')

# Encode target
le_target_adv = LabelEncoder()
y_encoded_adv = le_target_adv.fit_transform(y_adv)

# Encode remaining categorical features using One-Hot Encoding
existing_cat_cols = [col for col in categorical_cols if col in X_adv.columns]
X_adv = pd.get_dummies(X_adv, columns=existing_cat_cols, drop_first=True)

# Split into training and testing sets
X_train_adv, X_test_adv, y_train_adv, y_test_adv = train_test_split(X_adv, y_encoded_adv, test_size=0.2, random_state=42)

# --- 2. Hyperparameter Tuning using Grid Search ---
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

grid_search = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train_adv, y_train_adv)

# Get the best estimator
best_clf = grid_search.best_estimator_
print(f"Best Parameters: {grid_search.best_params_}\n")

# --- 3. Evaluate Optimized Model ---
y_pred_adv = best_clf.predict(X_test_adv)
accuracy_adv = accuracy_score(y_test_adv, y_pred_adv)
print(f"Optimized Model Accuracy: {accuracy_adv:.4f}\n")

print("--- Optimized Classification Report ---")
print(classification_report(y_test_adv, y_pred_adv, target_names=le_target_adv.classes_))
```

### 3. Training an Ensemble Model (Random Forest) for Higher Accuracy

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

# --- 1. Scale Numerical Features ---
scaler = StandardScaler()
X_train_scaled = X_train_adv.copy()
X_test_scaled = X_test_adv.copy()

# Scale only the numeric columns (columns that aren't dummy/one-hot columns)
non_dummy_cols = [col for col in numeric_candidates if col in X_train_adv.columns]
X_train_scaled[non_dummy_cols] = scaler.fit_transform(X_train_adv[non_dummy_cols])
X_test_scaled[non_dummy_cols] = scaler.transform(X_test_adv[non_dummy_cols])

# --- 2. Hyperparameter Tuning for Random Forest ---
rf_param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, None],
    'min_samples_split': [2, 5],
    'max_features': ['sqrt', 'log2', None]
}

rf_grid_search = GridSearchCV(RandomForestClassifier(random_state=42), rf_param_grid, cv=5, scoring='accuracy')
rf_grid_search.fit(X_train_scaled, y_train_adv)

best_rf = rf_grid_search.best_estimator_
print(f"Best Random Forest Parameters: {rf_grid_search.best_params_}\n")

# --- 3. Evaluate Optimized Random Forest ---
y_pred_rf = best_rf.predict(X_test_scaled)
accuracy_rf = accuracy_score(y_test_adv, y_pred_rf)
print(f"Random Forest Model Accuracy: {accuracy_rf:.4f}\n")

print("--- Random Forest Classification Report ---")
print(classification_report(y_test_adv, y_pred_rf, target_names=le_target_adv.classes_))
```

### 5. Food & Drink Search, Recommendation, and Opinion System

```python
import difflib

def search_and_recommend(user_query, df_menu, model, scaler, le_target, feature_cols):
    """
    Searches for a food/drink in the menu.
    If found, gives its details, ingredients, and an opinion using the model.
    If not found, suggests the most similar item available with ingredients.
    """
    products = df_menu['Products'].dropna().unique().tolist()

    # Find close matches
    matches = difflib.get_close_matches(user_query, products, n=1, cutoff=0.5)

    if matches and matches[0].lower() == user_query.lower():
        matched_item = matches[0]
        print(f"✓ Great news! '{matched_item}' is available in our menu.\n")
        provide_opinion(matched_item, df_menu, model, scaler, le_target, feature_cols)
    else:
        print(f"✗ Sorry, your searched item '{user_query}' is not directly available in our menu.")
        if matches:
            similar_item = matches[0]
            print(f"💡 However, we found a very similar item for you: '{similar_item}'")
            print("📖 Suggestion: You can order this or try its recipe/style as a substitute!\n")
            provide_opinion(similar_item, df_menu, model, scaler, le_target, feature_cols)
        else:
            # Default fallback if no close match is found at all
            fallback_item = products[0]
            print(f"💡 We couldn't find an exact match. You might like our popular option: '{fallback_item}'\n")
            provide_opinion(fallback_item, df_menu, model, scaler, le_target, feature_cols)

def provide_opinion(item_name, df_menu, model, scaler, le_target, feature_cols):
    # Extract row
    row = df_menu[df_menu['Products'] == item_name].iloc[0]

    # Get basic details
    energy = row.get('Total Energy', 'Unknown')
    sugar = row.get('Su gar', 'Unknown')
    fat = row.get('Tot al Fat', 'Unknown')
    allergens = row.get('Allergens', 'Unknown')

    # Use our trained RandomForest classifier to predict energy class dynamically
    item_encoded = X_adv[df_advanced['Products'] == item_name].iloc[0:1]

    if not item_encoded.empty:
        item_scaled = item_encoded.copy()
        non_dummy = [col for col in numeric_candidates if col in item_encoded.columns]
        item_scaled[non_dummy] = scaler.transform(item_encoded[non_dummy])

        # Predict class
        pred_code = model.predict(item_scaled)[0]
        pred_label = le_target.inverse_transform([pred_code])[0]
    else:
        pred_label = "Unknown"

    print(f"--- AI Opinion, Ingredients & Nutrition Profile for '{item_name}' ---")
    print(f"• Energy Content: {energy} kcal")
    print(f"• Sugar: {sugar}g | Total Fat: {fat}g")
    print(f"• Ingredients / Allergen Info: {allergens}")
    print(f"• AI Diet Analysis: This is classified as a *{pred_label} Energy* choice.")

    # AI Opinion Feedback
    if pred_label == 'High':
        print("💬 Opinion: This is a rich, high-energy option. Perfect for active days or post-workout fuel, but enjoy in moderation!")
    else:
        print("💬 Opinion: This is a lighter, low-energy option. Great for a healthy, guilt-free snack!")
    print("-" * 50)

# --- Demo Run ---
print("Demo 1: Searching for an existing item:")
search_and_recommend("Popcorn Med", df_advanced, best_rf, scaler, le_target_adv, X_adv.columns)

print("\nDemo 2: Searching for an item not in menu (suggesting similar option):")
search_and_recommend("Zinger Burger", df_advanced, best_rf, scaler, le_target_adv, X_adv.columns)
```
